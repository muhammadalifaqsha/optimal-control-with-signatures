```python
import base
import os    
os.environ['KMP_DUPLICATE_LIB_OK'] = 'True'

import numpy as np
import matplotlib.pyplot as plt
#from tqdm import tqdm_notebook as tqdm
from tqdm.auto import tqdm
import torch

#from data.gbm import GBM
import tensor_algebra as ta
import iisignature
import lqproblem_fbm_tracking as lqproblem
from fbm import FBM
import optimiser_lqproblem_fast as optimiser_lqproblem
from math import factorial
import pickle
from differint.differint import RL
```

```python
b0 = 0
b1 = 1
b2 = 1
kappa = 1
sigma0 = 1
sigma2 = 0
hurst_inv = 4

params = {'b0': b0, 'b1': b1, 'b2': b2,
          'sigma0': sigma0, 'sigma2': sigma2, 'X0': 0, 
          'A': 1, 'B': 1*kappa, 'C': 0, 'D':0, 'E':0, 'G': 0,
          'T': 1, 'Nt': 1_000, 'hurst_inv': hurst_inv}

with open(f'experiment_params_fbm_new.pkl', 'wb') as f:
    pickle.dump(params, f)
    f.close()

env = lqproblem.environment(**params)
env_optimal = lqproblem.environment(**params)
```

```python
nsims = 10_000
random_seed = 456

np.random.seed(random_seed)
env_optimal.simulate(nsims, lqproblem.strategy_optimal_feedback, order = None, fast = True)
```

```python
max_order_F = 3 # truncation of control tensor
max_order_x = 5 # truncation of state tensor
max_order_fn = max(2*max_order_x + 1, 2*max_order_F + 1) # total truncation order of cost functional
# poly_deg = 10

filename = f'lev_control_{max_order_F}_lev_state_{max_order_x}_fbm_analytical'
```

```python
dim = 2

optimiser_1 = optimiser_lqproblem.Optimiser(env, W_H_sig_expected, max_order_x = max_order_x, max_order_F = max_order_F)
# F = ta.sig_to_tensor(F, env.dim, order)
# optimiser_1.F = F
```

```python
U, V, p0 = optimiser_1.poly_coef(max_order_x = max_order_x)
F_flat = -(1/2)*np.linalg.inv(U)@V
```

```python
F = ta.sig_to_tensor(F_flat, env.dim, max_order_F)
optimiser_1.F.value = [F.value[n].clone().detach().requires_grad_() for n in range(max_order_F + 1)]
```

```python
with open(f'F_optimised_fbm_lqproblem_{filename}.pkl', 'wb') as f:
    pickle.dump(F.flatten(), f)
    f.close()
```

```python
strag_sig = lqproblem.sig_strag_dot_vec(F, order = max_order_F)
```

```python
np.random.seed(456)

env.simulate(nsims, strag_sig, max_order_F)
```
