```python
import base

import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm_notebook as tqdm
import torch

from data.gbm import GBM
import tensor_algebra as ta
import iisignature
import problem2
import optimiser_problem2
from math import factorial
import pickle
```

```python
sigma_s = 1e-1
b1 = 1e-3
b3 = 1e-2
b2 = 2
sigma2 = 2

params = {'b0': 1, 'b1': b1, 'b2': b2, 'b3': b3,
          'sigma0': 1, 'sigma2': sigma2, 'k': 1e-3, 
          'theta1': 10, 'theta2': 1e-2, 'S0': 100, 'Q0': 10, 'X0': 0, 
          'T': 0.10, 'Nt': 1_000}

env = problem2.environment(**params)
```

```python
np.random.seed(456)
dim = 2
order = 6

with open('W_sig_1.pkl', 'rb') as f:
    W_sig = pickle.load(f)
    f.close()

with open('F_optimised_problem2.pkl', 'rb') as f:
    F = pickle.load(f)
    f.close()
    
#env = problem2.environment(**params)
optimiser_1 = optimiser_problem2.Optimiser(env, W_sig, order = order)
F = ta.sig_to_tensor(F, optimiser_1.dim, optimiser_1.order)
```

```python
optimiser_1.F = F
```

```python
F
```

```python
F_new = ta.Tensor(dim = 2, order = 1)
```

```python
F_new.value[0][()] = 1
F_new.value[1][(0)] = 3
F_new.value[1][(1)] = 3

F_new_new = F_new

for i in range(order-1):
    F_new_new = F_new_new*F_new
```

```python
F_new_new
```

```python
optimiser_1.F = F_new_new
```

```python
# Define tensor algebra constant
#e0 = ta.word((), dim)
#e1 = ta.word((0,), dim)
#e2 = ta.word((1,), dim)

# Defining the representation of S and \hat{\alpha} in terms of the innovation process
trunc_lev = 7

P = optimiser_1.e0*optimiser_1.env.S0 + (optimiser_1.e0*optimiser_1.env.b0 + optimiser_1.env.b1*F)*optimiser_1.e1 + optimiser_1.e2*optimiser_1.env.sigma0
q = optimiser_1.e1*optimiser_1.env.b2 + optimiser_1.e2*optimiser_1.env.sigma2
r = optimiser_1.e1*optimiser_1.env.b3

max_iter = 400

exp_q = q.exp(MAX_ITER = max_iter, MAX_LEVEL = trunc_lev)
P_tilde = P*exp_q
r_tilde = r*exp_q

P_n = P_tilde
LF = P_n
M_max = F.order

LF_hist = [LF.truncate(trunc_lev)]

for i in range(trunc_lev+1):
    P_n = P_n.shuffle(F, max_order = trunc_lev)*r_tilde
    P_n = P_n.truncate(order = trunc_lev)
    LF += P_n
    LF = LF.truncate(trunc_lev)
    #LF_hist.append(LF)
    #if float(P_n.dot(P_n)) < 1e-8:
        #break
```
